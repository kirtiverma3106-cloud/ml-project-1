Univariate Analysis

```
# 1. Study Hours
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Study_Hours_Per_Day", kde=True)
plt.title("Distribution of Study Hours")
plt.show()
```

```
# 2. Sleep Hours
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Sleep_Hours_Per_Day", kde=True)
plt.title("Distribution of Sleep Hours")
plt.show()
```

```
# 3. GPA
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="GPA", kde=True)
plt.title("Distribution of GPA")
plt.show()
```

```
# 4. Stress Level
plt.figure(figsize=(7,5))
sns.countplot(data=df, x="Stress_Level")
plt.title("Distribution of Stress Levels")
plt.show()
```

Bivariate Analysis

```
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Study_Hours_Per_Day",
    y="GPA"
)

plt.title("Study Hours vs GPA")
plt.xlabel("Study Hours per Day")
plt.ylabel("GPA")
plt.show()
```

```
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Sleep_Hours_Per_Day",
    y="GPA"
)

plt.title("Sleep Hours vs GPA")
plt.xlabel("Sleep Hours per Day")
plt.ylabel("GPA")
plt.show()
```

```
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Physical_Activity_Hours_Per_Day",
    y="GPA"
)

plt.title("Physical Activity vs GPA")
plt.xlabel("Physical Activity Hours per Day")
plt.ylabel("GPA")
plt.show()
```

```
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Stress_Level",
    y="GPA"
)

plt.title("GPA Distribution Across Stress Levels")
plt.xlabel("Stress Level")
plt.ylabel("GPA")
plt.show()
```

Correlation analysis


```
# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation
correlation = numeric_df.corr()

# Display correlation matrix
correlation
```

```
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
